```python
def convertDate(timestamp):
  s = timestamp
  if ('AM' in s) or ('PM' in s):
    ts = time.mktime(datetime.datetime.strptime(s, "%b %d, %Y, %H:%M %p").timetuple()) 
    return datetime.datetime.fromtimestamp(ts).strftime("%Y-%m-%d %H:%M:%S")
  else:
    ts = time.mktime(datetime.datetime.strptime(s, "%d %b %Y, %H:%M").timetuple()) 
    return timestamp
```

```python
## Does contentizing
def contentize(content):
  content_dict = {} # create a dictionary for each comment section
  content_dict['user'] = user # create 41 as user id
  text = content.find("div", class_="_2pin").get_text() if content.find("div", class_="_2pin") else "None" # post text
  content_dict['text'] = text.replace('\n', ' ') # removes \n from post
  timestamp = content.find("div", class_="_3-94").get_text() # comments timestamp
  # content_dict['timestamp'] 
  # print(timestamp)
  content_dict['timestamp'] = convertDate(timestamp)
  # if "AM" or "PM" not in timestamp:
  #   new_timestamp = timestamp # otherwise if already 24 hours converted
  # else:
  #   new_timestamp = timestamp_split(timestamp) # convert to 24 hrs format

  print(content_dict)
  return content_dict


def remove_duplicate_posts(posts_list):
  ## For Removing Duplicate posts
  seen = set()
  new_posts_list = []
  for d in posts_list:
      t = tuple(d.items())
      if t not in seen:
          seen.add(t)
          new_posts_list.append(d)

  return new_posts_list
  # for i in new_posts_list:
  #   # print(i,'\n')


def remove_duplicate_group_comments(group_comments_list):
  ## For Removing Duplicate group comments  
  seen = set()
  new_group_comments_list = []
  for d in group_comments_list:
      t = tuple(d.items())
      if t not in seen:
          seen.add(t)
          new_group_comments_list.append(d)
  return new_group_comments_list
```
