Merged all 3 Codes Below

```python
# Opening the html file. If the file
# is present in different location, 
# exact location need to be mentioned
import re 

folders_list = ["comments/comments.html","posts/your_posts_1.html","groups/your_posts_and_comments_in_groups.html"]

for user in users_list: # goes through each user in the folder
  for folder in folders_list:
    filedir = f"/content/users/{user}/{folder}"
    # filedir = f"/content/users/{user}/comments/comments.html"
    HTMLFileToBeOpened = codecs.open(filedir, "r", "utf-8") # use codecs.open to open html files!
      
    # Reading the file and storing in a variable
    contents = HTMLFileToBeOpened.read()
      
    # Creating a BeautifulSoup object and
    # specifying the parser 
    soup = BeautifulSoup(contents, 'lxml')


    group_comments_list = []
    posts_list = []
    substring_group = 'Group:'

    # For Removing Group Names in posts:
    for span in soup("span"): # removes spans
      span.clear()
    for div in soup.find_all("div", {'class':'_3-95'}): # removes group names
        children = div.findChildren("div", recursive=False)
        for child in children:
          for child_div in child.find_all("div", {'class':'_3-95'}): # remove group names
            child_div.clear()


    for content in soup.find_all("div", class_="pam"): # content div of each comment
        # print(content,'\n')
        text = content.find("div", class_="_2pin").get_text() if content.find("div", class_="_2pin") else "None" # post text
        
        content_dict = {} # create a dictionary for each comment section
        content_dict['user'] = user # create 41 as user id
        content_dict['text'] = content.find("div", class_="_2pin").get_text() if content.find("div", class_="_2pin") else "None" # post text
        content_dict['text'] = content_dict['text'].replace('\n', ' ') # removes \n from post
        content_dict['text'] = re.sub("Updated.*$", "", content_dict['text']) # Removes the string Updated "timestamp" from texts
        content_dict['timestamp'] = content.find("div", class_="_3-94").get_text() # comments timestamp

        if substring_group in text: # if its a group comment
            # print('group')
            group_comments_list.append(content_dict)
        else:
            # print('NOT group')
            posts_list.append(content_dict)

    ## For Removing Duplicate posts
    seen = set()
    new_posts_list = []
    for d in posts_list:
        t = tuple(d.items())
        if t not in seen:
            seen.add(t)
            new_posts_list.append(d)

    for i in new_posts_list:
      print(i,'\n')

    ## For Removing Duplicate group comments  
    seen = set()
    new_group_comments_list = []
    for d in group_comments_list:
        t = tuple(d.items())
        if t not in seen:
            seen.add(t)
            new_group_comments_list.append(d)

    for i in new_group_comments_list:
      print(i,'\n')



    # for i in group_comments_list:
    #     print(i,'\n')

    # print('~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~')
    # for i in posts_list:
    #     print(i,'\n')
    # print('~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~')
```
