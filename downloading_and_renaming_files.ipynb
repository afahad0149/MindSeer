Downloading and extracting the files

```
#making a directory to download data
!mkdir /content/mindseer_data
```

```
#method 1:

def download_files(start, end):
    time.sleep(100)
    for i in range(start, end):
        path = '/content/mindseer_data/'
        participant_id = i
        url = file_links[i]
        file_id_index = url.find('id')
        file_id = url[file_id_index + 3 : ]
        data_zip = path + f'{participant_id}.zip'
        command = 'gdown --id ' + file_id + " -O " + data_zip
        m = os.system(command)
        print(m, participant_id)

        with zipfile.ZipFile(data_zip, "r") as zip_ref:
            path = path + f"{participant_id}/"
            zip_ref.extractall(path)
            zip_ref.close()
        
        #time.sleep(100)
```

```
#method 2:
import gdown

def download_files_2(start, end):
    for i in range(start, end):
        path = '/content/mindseer_data/'
        participant_id = i
        url = file_links[i]
        file_id_index = url.find('id')
        file_id = url[file_id_index + 3 : ]
        url = f"https://drive.google.com/uc?id={file_id}"
        data_zip = path + f'{participant_id}.zip'
        gdown.download(url, data_zip, quiet=False)

        with zipfile.ZipFile(data_zip, "r") as zip_ref:
            path = path + f"{participant_id}/"
            zip_ref.extractall(path)
            zip_ref.close()
```

```
download_files(0, 20)
```
